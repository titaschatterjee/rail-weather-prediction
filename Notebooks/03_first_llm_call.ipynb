```python
from dotenv import load_dotenv
import anthropic

load_dotenv("../.env")          # reads the key from .env into the environment
client = anthropic.Anthropic()  # picks up ANTHROPIC_API_KEY automatically

response = client.messages.create(
    model="claude-haiku-5-5",   # the smallest, cheapest model: ideal for learning
    max_tokens=300,
    messages=[{"role": "user",
               "content": "In 3 bullet points, how can heavy rain disrupt UK rail services?"}],
)

print(response.content[0].text)
print(response.usage)           # tokens in and out: what you pay for
```
